# Coding Challenge (Strings Only) – Ola Defect Analysis

# Scenario
You are a Data Analyst at Ola. You receive complaint data as a single string.
You must clean and analyze it using ONLY string operations.

# Input (Single String)
data = "Ride101-Ramesh-Late Arrival-Bangalore, Ride102-Suresh-rude behavior-Hyderabad,
Ride103-Mahesh-Late arrival-Bangalore"

```python
data = "Ride101-Ramesh-Late Arrival-Bangalore, Ride102-Suresh-rude behavior-Hyderabad, Ride103-Mahesh-Late arrival-Bangalore"
```

# Tasks (STRICTLY USING STRINGS)

# Task 1: Clean the Data
• Convert entire string to lowercase

```python
data = data.lower()
data
```

# Task 2: Count Total Complaints
• Count how many rides (records) are present

```python
print("Total Complaints: ",data.count("ride"))
```

# Task 3: Count "Late Arrival" Issues
• Count how many times "late arrival" appears

```python
print("Total Late Arrival issues:",data.count("late"))
```

# Task 4: Find Complaints from Bangalore
• Count how many times "bangalore" appears

```python
print("Total Complaints from Bangalore:",data.count("bangalore"))
```

# Task 5: Extract First Driver Name
• From first record → extract driver name

```python
print("First Driver Name:", data.split(",")[0].split("-")[1])
```

# Task 6: Replace Issue Names (Standardization)

• Replace all:
o "rude behavior" → "rude_behavior"
o "late arrival" → "late_arrival"

```python
data = data.replace("rude behavior","rude_behavior")
data = data.replace("late arrival","late_arrival")
data
```
